### Calculate fingerprint descriptors

```python
from padelpy import padeldescriptor
from tqdm import tqdm
import threading
import time
import pandas as pd

num_molecules = 206

def run_padel():
    padeldescriptor(
        fingerprints=True,
        mol_dir=smi_file,
        d_file=output_csv,
        threads=2
    )

padel_thread = threading.Thread(target=run_padel)
padel_thread.start()

with tqdm(total=num_molecules, desc="PaDEL Progress") as pbar:
    last_count = 0
    while padel_thread.is_alive():
        try:
            df = pd.read_csv(output_csv)
            current_count = len(df) - 1
        except FileNotFoundError:
            current_count = 0
        delta = current_count - last_count
        if delta > 0:
            pbar.update(delta)
            last_count = current_count
        time.sleep(1)
    try:
        df = pd.read_csv(output_csv)
        final_count = len(df) - 1
    except FileNotFoundError:
        final_count = 0
    pbar.update(final_count - last_count)

padel_thread.join()
print(f"PaDEL descriptors saved to: {output_csv}")
```
